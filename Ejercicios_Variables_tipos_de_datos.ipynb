Calculadora Multiformato Desarrolla un programa que permita al usuario ingresar dos números y, a partir de ellos, realice diversas operaciones matemáticas. El sistema debe procesar la suma, la diferencia, el producto, la división, el residuo y la potencia, mostrando cada resultado junto a su respectivo formato numérico. Además, la solución debe encargarse de transformar la información recibida para asegurar que los cálculos se efectúen con precisión y se presenten de manera ordenada

```python
x= int(input("Ingrese el primer numero: "))
y= int(input("Ingrese el segundo numero: "))
print("La suma es: ", x+y)
print("La diferencia es: ", x-y)
print("El producto es: ", x*y)
print("La division es:", x/y)
print("El residuo es: ", x%y)
print("La potencia es: ", x**y)
```

Conversión entre Sistemas Numéricos El objetivo es construir una aplicación que solicite un número en formato decimal y lo convierta a sus equivalentes en otros sistemas, como el binario, el octal y el hexadecimal. Asimismo, el programa deberá permitir realizar la conversión inversa, de modo que el usuario pueda ingresar un número en cualquiera de estos formatos y obtener su valor en el sistema decimal. Se espera que la solución organice y formatee los resultados de manera clara para facilitar la comparación entre las distintas representaciones.

```python
decimal= int(input("ingrese un numero decimal:"))
binario= bin(decimal)
octal= oct(decimal)
hexadecimal= hex(decimal)
print("decimal:", decimal)
print("binario:", binario)
print("octal:", octal)
print("hexadecimal:", hexadecimal)
```

Control de Precios y Descuentos Imagina una solución que pida al usuario el nombre de un producto junto con su precio original. El reto consiste en calcular un descuento fijo, mostrar el valor descontado y determinar el precio final tras aplicar la reducción. La información se debe presentar de forma legible, utilizando formatos que permitan distinguir claramente los montos involucrados y garantizando la exactitud en los cálculos mediante operaciones matemáticas adecuadas

```python
producto = input("Ingrese el nombre del producto: ")
precio_original = int(input("Ingrese el precio original: "))
descuento = int(input("Ingrese el descuento en porcentaje (%): "))
descuento_aplicado = precio_original * (descuento / 100)
total_a_pagar = precio_original - descuento_aplicado

print("\nResumen de la compra:")
print("Producto:", producto)
print(f"Precio original: ${precio_original:.2f}")
print(f"Descuento aplicado: ${descuento_aplicado:.2f} ({descuento:.0f}%)")
print(f"Total a pagar: ${total_a_pagar:.2f}")
```

Análisis Detallado de una Frase Crea una aplicación que invite al usuario a introducir una frase y que, a partir de ella, extraiga distintos elementos: la longitud total, la letra inicial y final, y una sección intermedia del texto. El sistema debe además transformar el contenido a diferentes presentaciones, como mayúsculas y minúsculas, y contar cuántas veces aparece un carácter específico. Este ejercicio pone a prueba la capacidad de manipular y transformar secuencias textuales, así como de realizar comparaciones y conteos precisos.

```python
frase= input("Ingrese una frase:")
print("La longitud de la frase es:", len(frase))
print("La letra inicial es:", frase[0])
print("La letra final es:", frase[-1])
print("La frase invertida es:", frase[::-1])
print("La frase en mayúsculas es:", frase.upper())
print("La frase en minúsculas es:", frase.lower())
print(frase)
```

Cálculo de Bonificaciones Laborales Diseña una solución que reciba datos de un trabajador, tales como su remuneración actual y el tiempo que ha estado en la empresa, para calcular una bonificación basada en un porcentaje determinado. El programa debe computar el incremento correspondiente, mostrar el monto adicional y el nuevo total, presentando la información de forma clara y formateada. El desafío consiste en aplicar operaciones aritméticas precisas y transformar los datos de entrada para lograr un resultado correcto y comprensible.

```python
nombre = input("Ingrese el nombre del trabajador: ")
apellido = input("Ingrese el apellido del trabajador: ")
salario = int(input("Ingrese su salario: "))
meses = int(input("Ingrese los meses que ha estado en la empresa: "))


if meses > 12:
    bonificacion = salario * 0.5
elif 6 < meses <= 12:
    bonificacion = salario * 0.3
elif 3 <= meses <= 6:
    bonificacion = salario * 0.1
else:
    bonificacion = 0


salario_total = salario + bonificacion


print(nombre)
print(apellido)
print(f"Salario base: ${salario:.2f}")
print(f"Meses en la empresa: {meses}")
print(f"Bonificación: ${bonificacion:.2f}")
print(f"Salario total con bonificación: ${salario_total:.2f}")

```
